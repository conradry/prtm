```python
from prtm.ligand import load_molecule_from_file
from prtm.models import DiffDockForLigandDocking
from prtm.protein import Protein14
from prtm.visual import view_molecule, view_structure_with_ligand, view_ligand_docking_trajectory
```

```python
docker = DiffDockForLigandDocking()
```

```python
example_code = "6o5u" # 1a46, 6ahs, 6moa, 6o5u, 6w70
protein_file = f"data/{example_code}_protein_processed.pdb"
ligand_file = f"data/{example_code}_ligand.sdf"

with open(protein_file, mode="r") as f:
    protein_pdb_str = f.read()
structure = Protein14.from_pdb_string(protein_pdb_str)

ligand = load_molecule_from_file(ligand_file)
```

```python
# 3D view of the ligand
view_molecule(ligand)
```

```python
poses, aux_output = docker(structure, ligand)
```

```python
# Show the docking trajectories
view_ligand_docking_trajectory(structure, aux_output["pdb_trajectories"], surface_opacity=0.7)
```

```python
# Show final poses for all ligands
view_structure_with_ligand(structure, poses)
```
